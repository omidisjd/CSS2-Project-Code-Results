```python
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    # Makes CUDA more deterministic
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
```

```python
class ClinicalPreprocessor:
    @staticmethod
    def remove_hair(img):
        # Convert to grayscale
        gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
        # Blackhat filter to identify hair-like structures
        kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (17, 17))
        blackhat = cv2.morphologyEx(gray, cv2.MORPH_BLACKHAT, kernel)
        # Create a mask for hair
        _, threshold = cv2.threshold(blackhat, 10, 255, cv2.THRESH_BINARY)
        # Inpaint the original image based on the mask
        final_img = cv2.inpaint(img, threshold, 1, cv2.INPAINT_TELEA)
        return final_img

    @staticmethod
    def apply_clahe(img):
        # Convert to LAB color space to apply CLAHE on the Lightness channel
        lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
        l, a, b = cv2.split(lab)
        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
        cl = clahe.apply(l)
        limg = cv2.merge((cl, a, b))
        return cv2.cvtColor(limg, cv2.COLOR_LAB2RGB)

    @classmethod
    def process(cls, img_path, track='robust'):
        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
        
        if track == 'robust':
            img = cls.remove_hair(img)
            img = cls.apply_clahe(img)
            
        return img
```

```python
class StressTest:
    @staticmethod
    def add_gaussian_blur(img, level=1):
        # level 1-5 maps to kernel size
        k_size = (level * 2) + 1
        return cv2.GaussianBlur(img, (k_size, k_size), 0)

    @staticmethod
    def add_salt_pepper(img, prob=0.05):
        output = np.copy(img)
        # Salt
        num_salt = np.ceil(prob * img.size * 0.5)
        coords = [np.random.randint(0, i - 1, int(num_salt)) for i in img.shape]
        output[tuple(coords)] = 255
        # Pepper
        num_pepper = np.ceil(prob * img.size * 0.5)
        coords = [np.random.randint(0, i - 1, int(num_pepper)) for i in img.shape]
        output[tuple(coords)] = 0
        return output

    @staticmethod
    def add_occlusion(img, level=1):
        # Simulates a finger or hair edge blocking part of the lesion
        h, w, _ = img.shape
        output = img.copy()
        size = int(w * (level * 0.1))
        x = np.random.randint(0, w - size)
        y = np.random.randint(0, h - size)
        output[y:y+size, x:x+size] = 0 # Black box occlusion
        return output
```

```python
df = pd.read_csv(CSV_PATH)
# Stratified Split (80% Train, 10% Val, 10% Test)
train_df, test_df = train_test_split(df, test_size=0.2, stratify=df['dx'], random_state=SEED)
val_df, test_df = train_test_split(test_df, test_size=0.5, stratify=test_df['dx'], random_state=SEED)

# Create label mapping
label_map = {cat: i for i, cat in enumerate(CATEGORIES)}

# Convert diagnosis labels into numeric labels for model training and evaluation
train_df['label_idx'] = train_df['dx'].map(label_map)
val_df['label_idx'] = val_df['dx'].map(label_map)
test_df['label_idx'] = test_df['dx'].map(label_map)
```

```python
class HAM10000Dataset(Dataset):
    def __init__(self, dataframe, root_dir, transform=None, track='baseline'):
        """
        Args:
            dataframe (Pandas DataFrame): Contains 'image_id' and 'label_idx'.
            root_dir (string): Directory with all the raw images.
            transform (callable, optional): PyTorch transforms to apply.
            track (string): 'baseline' or 'robust' to dictate preprocessing.
        """
        self.df = dataframe.reset_index(drop=True)
        self.root_dir = root_dir
        self.transform = transform
        self.track = track
        
        # Instantiate your preprocessing class here
        self.preprocessor = ClinicalPreprocessor() 

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        # 1. Get image path and label
        img_id = self.df.iloc[idx]['image_id']
        label = self.df.iloc[idx]['label_idx']
        img_path = os.path.join(self.root_dir, f"{img_id}.jpg")
        
        # 2. On-the-fly Clinical Preprocessing (OpenCV)
        # Baseline = Resize only; Robust = Hair Removal + CLAHE + Resize
        img_array = self.preprocessor.process(img_path, track=self.track)
        
        # 3. Convert NumPy array back to PIL Image for torchvision transforms
        image = Image.fromarray(img_array.astype('uint8'))
        
        # 4. Apply PyTorch Augmentations & Normalization
        if self.transform:
            image = self.transform(image)
            
        return image, torch.tensor(label, dtype=torch.long)

print(f"Dataset Split Complete: {len(train_df)} Train, {len(val_df)} Val, {len(test_df)} Test")
```

```python
class StressTestDataset(HAM10000Dataset):
    def __init__(self, dataframe, root_dir, noise_type=None, intensity=1, **kwargs):
        # Initialize the base HAM10000 dataset with any extra arguments
        super().__init__(dataframe, root_dir, **kwargs)

        # Store the type of image degradation to apply
        self.noise_type = noise_type

        # Store the severity level of the degradation
        self.intensity = intensity

        # Create stress testing tool for applying image corruptions
        self.stress_tool = StressTest()

    def __getitem__(self, idx):
        # Get the original image and label using the parent dataset logic
        image, label = super().__getitem__(idx)

        # Rebuild image path for the current sample
        img_path = os.path.join(self.root_dir, f"{self.df.iloc[idx]['image_id']}.jpg")

        # Apply clinical preprocessing before adding stress corruption
        raw_img = self.preprocessor.process(img_path, track=self.track)

        # Apply Gaussian blur if selected
        if self.noise_type == 'blur':
            raw_img = self.stress_tool.add_gaussian_blur(raw_img, level=self.intensity)

        # Apply salt-and-pepper noise if selected
        elif self.noise_type == 's&p':
            raw_img = self.stress_tool.add_salt_pepper(raw_img, prob=0.02 * self.intensity)

        # Apply black-box occlusion if selected
        elif self.noise_type == 'occlusion':
            raw_img = self.stress_tool.add_occlusion(raw_img, level=self.intensity)

        # Apply combined corruption for the hardest stress level
        elif self.noise_type == 'combined':
            raw_img = self.stress_tool.add_gaussian_blur(raw_img, level=self.intensity)
            raw_img = self.stress_tool.add_salt_pepper(raw_img, prob=0.02 * self.intensity)
            raw_img = self.stress_tool.add_occlusion(raw_img, level=self.intensity)

        # Convert NumPy image array back to PIL image for transforms
        image = Image.fromarray(raw_img.astype('uint8'))

        # Apply torchvision transforms if provided
        if self.transform:
            image = self.transform(image)

        # Return stressed image and original label
        return image, label
```

```python
# Track 1: Baseline Augmentation (Minimal)
basic_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])

# Track 2: Robust Augmentation (Heavy)
# Focuses on color constancy and geometric invariance
class AddGaussianNoise:
    def __init__(self, mean=0., std=0.02):
        self.mean = mean
        self.std = std

    def __call__(self, tensor):
        return tensor + torch.randn_like(tensor) * self.std

aug_transforms = transforms.Compose([
    transforms.Resize((224, 224)),

    # Mild viewpoint changes
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(20),
    transforms.RandomAffine(
        degrees=0,
        translate=(0.05, 0.05),
        scale=(0.95, 1.05)
    ),

    # Smartphone sensor effects
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.25,
        saturation=0.2,
        hue=0.05
    ),

    # Optical imperfections
    transforms.GaussianBlur(kernel_size=3, sigma=(0.1, 1.2)),

    transforms.ToTensor(),

    # Sensor noise
    AddGaussianNoise(std=0.015),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# Validation Transforms have No augmentation, just normalization, same as basic_transforms


# --- Instantiate DataLoaders ---
BATCH_SIZE = 32
NUM_WORKERS = 4 # Adjust based on CPU cores to speed up on-the-fly processing

# --- 1. Train Loaders (The Dual-Track) ---
train_loader_base = DataLoader(
    HAM10000Dataset(train_df, BASE_DIR, transform=basic_transforms, track='baseline'), 
    batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS
)

train_loader_robust = DataLoader(
    HAM10000Dataset(train_df, BASE_DIR, transform=aug_transforms, track='robust'), 
    batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS
)

# --- 2. Validation Loader (The Benchmark) ---
val_loader_base = DataLoader(
    HAM10000Dataset(val_df, BASE_DIR, transform=basic_transforms, track='baseline'),
    batch_size=32, shuffle=False
)

val_loader_robust = DataLoader(
    HAM10000Dataset(val_df, BASE_DIR, transform=basic_transforms, track='robust'),
    batch_size=32, shuffle=False
)

# --- 3. Test Loaders (The Evaluation) ---
# A standard clean test loader to get your baseline metrics
test_loader_clean = DataLoader(
    HAM10000Dataset(test_df, BASE_DIR, transform=basic_transforms, track='baseline'), 
    batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS
)


# to dynamically inject Gaussian Blur, Salt & Pepper, etc.
def get_stress_test_loader(
    dataframe,
    noise_type=None,
    level=0,
    track='baseline'
):
    """
    Creates a test DataLoader for clean, augmented, or stressed evaluation.

    Level 0:
        Clean test images.
        No heavy augmentation.
        No synthetic noise.

    Level 1:
        Heavy smartphone-style augmentation only.
        No synthetic noise.

    Levels 2-4:
        Heavy smartphone-style augmentation + one selected synthetic noise type.
        Noise strength increases with level.

    Level 5:
        Heavy smartphone-style augmentation + blur + salt-and-pepper + occlusion.
        Uses intensity 3 for each corruption.
    """

    if level == 0:
        # LEVEL 0: Clean test set
        test_dataset = HAM10000Dataset(
            dataframe,
            root_dir=BASE_DIR,
            transform=basic_transforms,
            track=track
        )

    elif level == 1:
        # LEVEL 1: Heavy smartphone-style augmentation only
        test_dataset = HAM10000Dataset(
            dataframe,
            root_dir=BASE_DIR,
            transform=aug_transforms,
            track=track
        )

    elif level in [2, 3, 4]:
        # LEVELS 2-4: One synthetic noise type with increasing intensity
        if noise_type is None:
            raise ValueError("noise_type must be provided for levels 2, 3, 4 and 5.")

        # Map stress level to noise intensity
        noise_intensity = level - 1

        test_dataset = StressTestDataset(
            dataframe,
            root_dir=BASE_DIR,
            noise_type=noise_type,
            intensity=noise_intensity,
            transform=aug_transforms,
            track=track
        )

    elif level == 5:
        # LEVEL 5: Combined corruption using max intensity
        test_dataset = StressTestDataset(
            dataframe,
            root_dir=BASE_DIR,
            noise_type='combined',
            intensity=3,
            transform=aug_transforms,
            track=track
        )

    else:
        raise ValueError("level must be one of: 0, 1, 2, 3, 4, 5")

    return DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS
    )
```

```python
import torch.nn as nn
import timm # High-quality library for SOTA models

def build_model(model_name, num_classes=7):
    if model_name == 'resnet50':
        # Convolutional Baseline
        model = timm.create_model('resnet50', pretrained=True, num_classes=num_classes)
        
    elif model_name == 'vit_b16':
        # State-of-the-art Self-Attention
        model = timm.create_model('vit_base_patch16_224', pretrained=True, num_classes=num_classes)
        
    elif model_name == 'efficientnet_b0':
        # Mobile-optimized architecture
        model = timm.create_model('efficientnet_b0', pretrained=True, num_classes=num_classes)
    
    else:
        raise ValueError("Model name not supported.")
        
    return model

# Initialize the trio
models = {
    "ResNet-50": build_model('resnet50'),
    "ViT-B/16": build_model('vit_b16'),
    "EfficientNet-B0": build_model('efficientnet_b0')
}
```

```python
def train_one_epoch(model, dataloader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    
    for images, labels in dataloader:
        images, labels = images.to(device), labels.to(device)
        
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
        
    return running_loss / len(dataloader), 100. * correct / total

def validate(model, dataloader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    
    with torch.no_grad():
        for images, labels in dataloader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            
            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
            
    return running_loss / len(dataloader), 100. * correct / total
```

```python
def compute_class_weights(df, num_classes=7):
    counts = df['label_idx'].value_counts().to_dict()
    weights = np.zeros(num_classes, dtype=np.float32)

    for cls in range(num_classes):
        weights[cls] = 1.0 / counts[cls]

    weights = weights / weights.sum() * num_classes
    return torch.tensor(weights, dtype=torch.float32)
```

```python
def make_safe_name(model_name):
    return model_name.replace("-", "_").replace("/", "_")


def get_checkpoint_path(model_name, track):
    safe_name = make_safe_name(model_name)
    track_lower = track.lower()
    return os.path.join(SAVE_DIR, f"best_{safe_name}_{track_lower}.pth")
```

```python
def load_trained_model(model_name, track, device):
    model_key = model_keys[model_name]
    model = build_model(model_key, num_classes=NUM_CLASSES)

    checkpoint_path = get_checkpoint_path(model_name, track)

    if not os.path.exists(checkpoint_path):
        raise FileNotFoundError(f"Checkpoint not found: {checkpoint_path}")

    state_dict = torch.load(checkpoint_path, map_location=device)
    model.load_state_dict(state_dict)

    model = model.to(device)
    model.eval()

    return model
```

```python
def save_all_variants_grid(
    img_path,
    save_path="all_preprocessing_variants.png",
    seed=42
):
    # Reproducibility
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    preprocessor = ClinicalPreprocessor()
    stress_tool = StressTest()

    images = []
    titles = []
    groups = []

    # ----------------------------
    # TRAINING METHODS
    # ----------------------------

    # Baseline training preprocessing: resize only
    baseline_img = preprocessor.process(img_path, track="baseline")
    images.append(baseline_img)
    titles.append("Baseline training\nResize only")
    groups.append("Training methods")

    # Robust training preprocessing: hair removal + CLAHE
    robust_img = preprocessor.process(img_path, track="robust")
    images.append(robust_img)
    titles.append("Robust training\nHair removal + CLAHE")
    groups.append("Training methods")

    # Visual version of robust training augmentation
    visual_aug = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(20),
        transforms.RandomAffine(
            degrees=0,
            translate=(0.05, 0.05),
            scale=(0.95, 1.05)
        ),
        transforms.ColorJitter(
            brightness=0.2,
            contrast=0.25,
            saturation=0.2,
            hue=0.05
        ),
        transforms.GaussianBlur(kernel_size=3, sigma=(0.1, 1.2))
    ])

    # Robust augmented training example
    aug_img = np.array(
        visual_aug(Image.fromarray(robust_img.astype("uint8")))
    )
    images.append(aug_img)
    titles.append("Robust training\nSmartphone augmentation")
    groups.append("Training methods")

    # ----------------------------
    # TESTING METHODS
    # ----------------------------

    # Level 0: clean test image
    images.append(baseline_img)
    titles.append("Test level 0\nClean image")
    groups.append("Testing / stress-test methods")

    # Level 1: smartphone-style augmented test image
    images.append(aug_img)
    titles.append("Test level 1\nSmartphone augmentation")
    groups.append("Testing / stress-test methods")

    # Levels 2-4: individual stress corruptions
    noise_types = ["blur", "s&p", "occlusion"]

    for noise_type in noise_types:
        for level in [2, 3, 4]:
            intensity = level - 1
            noisy = aug_img.copy()

            if noise_type == "blur":
                noisy = stress_tool.add_gaussian_blur(noisy, level=intensity)

            elif noise_type == "s&p":
                noisy = stress_tool.add_salt_pepper(
                    noisy,
                    prob=0.02 * intensity
                )

            elif noise_type == "occlusion":
                noisy = stress_tool.add_occlusion(noisy, level=intensity)

            images.append(noisy)
            titles.append(f"Test level {level}\n{noise_type}")
            groups.append("Testing / stress-test methods")

    # Level 5: combined corruption
    combined = aug_img.copy()
    combined = stress_tool.add_gaussian_blur(combined, level=3)
    combined = stress_tool.add_salt_pepper(combined, prob=0.06)
    combined = stress_tool.add_occlusion(combined, level=3)

    images.append(combined)
    titles.append("Test level 5\nCombined corruption")
    groups.append("Testing / stress-test methods")

    # ----------------------------
    # Plot grouped grid
    # ----------------------------

    unique_groups = ["Training methods", "Testing / stress-test methods"]

    cols = 4
    total_rows = 0

    # Work out how many rows are needed for each group
    for group in unique_groups:
        group_count = groups.count(group)
        total_rows += int(np.ceil(group_count / cols)) + 1

    plt.figure(figsize=(4 * cols, 4 * total_rows))

    plot_idx = 1

    for group in unique_groups:
        group_items = [
            (img, title)
            for img, title, g in zip(images, titles, groups)
            if g == group
        ]

        # Add group heading
        ax = plt.subplot(total_rows, cols, plot_idx)
        ax.text(
            0.5,
            0.5,
            group,
            fontsize=16,
            fontweight="bold",
            ha="center",
            va="center"
        )
        ax.axis("off")

        # Fill the rest of the heading row with empty plots
        for _ in range(cols - 1):
            plot_idx += 1
            ax = plt.subplot(total_rows, cols, plot_idx)
            ax.axis("off")

        plot_idx += 1

        # Plot all images in the current group
        for img, title in group_items:
            plt.subplot(total_rows, cols, plot_idx)
            plt.imshow(img)
            plt.title(title, fontsize=9)
            plt.axis("off")
            plot_idx += 1

        # Move to the next full row if needed
        while (plot_idx - 1) % cols != 0:
            ax = plt.subplot(total_rows, cols, plot_idx)
            ax.axis("off")
            plot_idx += 1

    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"Saved grouped preprocessing + stress-test grid to: {save_path}")
```

```python
from sklearn.metrics import recall_score

DEADLY_CLASS_NAME = "mel" #melanoma

def compute_deadly_class_recall(model, dataloader, deadly_class_idx, device):
    model.eval()

    all_preds = []
    all_labels = []

    with torch.no_grad():
        for images, labels in dataloader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            preds = outputs.argmax(dim=1)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    all_preds = np.array(all_preds)
    all_labels = np.array(all_labels)

    # Binary recall: deadly vs not-deadly
    y_true = (all_labels == deadly_class_idx).astype(int)
    y_pred = (all_preds == deadly_class_idx).astype(int)

    recall = recall_score(y_true, y_pred, zero_division=0)

    return recall
```

```python
def evaluate_deadly_class_recall(
    test_df,
    deadly_class_name,
    noise_type=None,
    level=0
):
    deadly_class_idx = label_map[deadly_class_name]

    results = []

    for track in tracks:
        track_arg = track.lower()

        for model_name in model_keys.keys():
            print(f"Evaluating {model_name} | {track} | level {level}")

            model = load_trained_model(
                model_name=model_name,
                track=track,
                device=device
            )

            loader = get_stress_test_loader(
                dataframe=test_df,
                noise_type=noise_type,
                level=level,
                track=track_arg
            )

            recall = compute_deadly_class_recall(
                model=model,
                dataloader=loader,
                deadly_class_idx=deadly_class_idx,
                device=device
            )

            results.append({
                "model": model_name,
                "track": track,
                "noise_type": noise_type if noise_type else "clean",
                "level": level,
                "deadly_class": deadly_class_name,
                "recall": recall
            })

    return pd.DataFrame(results)
```

```python
# Simple Grad-CAM implementation using hooks
class SimpleGradCAM:
    def __init__(self, model, target_layer):
        # Store model and target convolutional layer
        self.model = model
        self.target_layer = target_layer

        # Store activations and gradients from target layer
        self.activations = None
        self.gradients = None

        # Register hooks to capture forward activations and backward gradients
        self.forward_hook = target_layer.register_forward_hook(self.save_activation)
        self.backward_hook = target_layer.register_full_backward_hook(self.save_gradient)

    def save_activation(self, module, input, output):
        # Save feature maps from forward pass
        self.activations = output.detach()

    def save_gradient(self, module, grad_input, grad_output):
        # Save gradients from backward pass
        self.gradients = grad_output[0].detach()

    def generate(self, input_tensor, target_class=None):
        # Put model in evaluation mode
        self.model.eval()

        # Forward pass
        output = self.model(input_tensor)

        # If no class is specified, use predicted class
        if target_class is None:
            target_class = output.argmax(dim=1).item()

        # Clear old gradients
        self.model.zero_grad()

        # Backward pass for the target class score
        class_score = output[:, target_class]
        class_score.backward()

        # Average gradients across spatial dimensions
        weights = self.gradients.mean(dim=(2, 3), keepdim=True)

        # Weighted combination of activation maps
        cam = (weights * self.activations).sum(dim=1).squeeze()

        # Keep only positive importance
        cam = torch.relu(cam)

        # Normalize heatmap to range 0-1
        cam = cam.cpu().numpy()
        cam = cam - cam.min()
        cam = cam / (cam.max() + 1e-8)

        return cam, target_class

    def remove_hooks(self):
        # Remove hooks when Grad-CAM object is no longer needed
        self.forward_hook.remove()
        self.backward_hook.remove()
```

```python
def denormalize_tensor(img_tensor):
    # ImageNet normalization values used in your transforms
    mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

    # Reverse normalization
    img = img_tensor.cpu() * std + mean

    # Clamp values to valid image range
    img = torch.clamp(img, 0, 1)

    # Convert tensor from [C, H, W] to [H, W, C]
    img = img.permute(1, 2, 0).numpy()

    return img


def overlay_cam_on_image(rgb_img, cam):
    # Resize CAM to match image size
    cam_resized = cv2.resize(cam, (rgb_img.shape[1], rgb_img.shape[0]))

    # Convert heatmap to RGB color map
    heatmap = cv2.applyColorMap(np.uint8(255 * cam_resized), cv2.COLORMAP_JET)
    heatmap = cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB)
    heatmap = heatmap.astype(np.float32) / 255.0

    # Blend original image and heatmap
    overlay = 0.55 * rgb_img + 0.45 * heatmap

    # Keep values in valid range
    overlay = np.clip(overlay, 0, 1)

    return overlay


def get_resnet_target_layer(model):
    # Last convolutional block of ResNet-50 for Grad-CAM
    return model.layer4[-1]
```

```python
def run_gradcam_on_sample(model, dataset, idx, device, class_names=None):
    # Get image tensor and true label from dataset
    image_tensor, label = dataset[idx]

    # Add batch dimension and move to device
    input_tensor = image_tensor.unsqueeze(0).to(device)

    # Get model prediction
    model.eval()
    with torch.no_grad():
        output = model(input_tensor)
        pred_class = output.argmax(dim=1).item()
        confidence = torch.softmax(output, dim=1)[0, pred_class].item()

    # Select target layer for ResNet-50
    target_layer = get_resnet_target_layer(model)

    # Create Grad-CAM object
    gradcam = SimpleGradCAM(model, target_layer)

    # Generate CAM for predicted class
    cam, target_class = gradcam.generate(input_tensor, target_class=pred_class)

    # Remove hooks after use
    gradcam.remove_hooks()

    # Convert image tensor back to viewable RGB image
    rgb_img = denormalize_tensor(image_tensor)

    # Overlay CAM on image
    overlay = overlay_cam_on_image(rgb_img, cam)

    # Create readable labels
    label_val = label.item() if isinstance(label, torch.Tensor) else label
    true_name = class_names[label_val] if class_names else str(label_val)
    pred_name = class_names[pred_class] if class_names else str(pred_class)

    return {
        "image": rgb_img,
        "cam_overlay": overlay,
        "true_label": label_val,
        "pred_label": pred_class,
        "true_name": true_name,
        "pred_name": pred_name,
        "confidence": confidence,
        "correct": pred_class == label_val
    }
```

```python
def find_correct_and_incorrect_examples(model, dataset, device, max_samples=300):
    # Store first correct and first incorrect example found
    correct_idx = None
    incorrect_idx = None

    model.eval()

    # Search through dataset samples
    for idx in range(min(len(dataset), max_samples)):
        image_tensor, label = dataset[idx]
        input_tensor = image_tensor.unsqueeze(0).to(device)

        with torch.no_grad():
            output = model(input_tensor)
            pred = output.argmax(dim=1).item()

        # Save first correct example
        
        label_val = label.item() if isinstance(label, torch.Tensor) else label
        if pred == label_val and correct_idx is None:
            correct_idx = idx

        # Save first incorrect example
        if pred != label_val and incorrect_idx is None:
            incorrect_idx = idx

        # Stop once both are found
        if correct_idx is not None and incorrect_idx is not None:
            break

    return correct_idx, incorrect_idx
```

```python
def find_robust_improvement_example(
    baseline_model,
    robust_model,
    dataset_baseline,
    dataset_robust,
    device,
    max_samples=2000
):
    for idx in range(min(len(dataset_baseline), max_samples)):

        base_img, label = dataset_baseline[idx]
        robust_img, _ = dataset_robust[idx]

        label_val = label.item() if isinstance(label, torch.Tensor) else label

        base_input = base_img.unsqueeze(0).to(device)
        robust_input = robust_img.unsqueeze(0).to(device)

        baseline_model.eval()
        robust_model.eval()

        with torch.no_grad():
            base_pred = baseline_model(base_input).argmax(dim=1).item()
            robust_pred = robust_model(robust_input).argmax(dim=1).item()

        # ✅ Correct condition
        if base_pred != label_val and robust_pred == label_val:
            return idx

    return None
```

```python
def save_gradcam_figure(results, save_path, title):
    # Create a figure with original image and Grad-CAM overlay
    plt.figure(figsize=(8, 4))

    # Original image
    plt.subplot(1, 2, 1)
    plt.imshow(results["image"])
    plt.title("Original image")
    plt.axis("off")

    # Grad-CAM overlay
    plt.subplot(1, 2, 2)
    plt.imshow(results["cam_overlay"])
    plt.title(
        f"Grad-CAM\nTrue: {results['true_name']} | Pred: {results['pred_name']}\n"
        f"Confidence: {results['confidence']:.2f}"
    )
    plt.axis("off")

    # Add main title
    plt.suptitle(title, fontsize=13)

    # Save and display
    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"Saved Grad-CAM figure to: {save_path}")
```

```python
def get_shared_stress_image(img_path, preprocessor, stress_tool):
    """
    Applies the SAME preprocessing + stress corruption once,
    then returns that image so both models use identical input.
    """

    # Load + preprocess ONCE (use robust to get cleanest version)
    base_img = preprocessor.process(img_path, track="robust")

    # Apply SAME combined level-5 corruption (fixed intensity)
    stressed = stress_tool.add_gaussian_blur(base_img, level=3)
    stressed = stress_tool.add_salt_pepper(stressed, prob=0.06)
    stressed = stress_tool.add_occlusion(stressed, level=3)

    return base_img, stressed
```

```python
def run_gradcam_on_tensor(model, image_tensor, device, class_names=None):

    input_tensor = image_tensor.unsqueeze(0).to(device)

    model.eval()
    with torch.no_grad():
        output = model(input_tensor)
        pred_class = output.argmax(dim=1).item()
        confidence = torch.softmax(output, dim=1)[0, pred_class].item()

    target_layer = get_resnet_target_layer(model)
    gradcam = SimpleGradCAM(model, target_layer)

    cam, _ = gradcam.generate(input_tensor, target_class=pred_class)
    gradcam.remove_hooks()

    rgb_img = denormalize_tensor(image_tensor)
    overlay = overlay_cam_on_image(rgb_img, cam)

    pred_name = class_names[pred_class] if class_names else str(pred_class)

    return {
        "image": rgb_img,
        "cam_overlay": overlay,
        "pred_label": pred_class,
        "pred_name": pred_name,
        "confidence": confidence
    }
```

```python
def plot_combined_comparison(results_df):
    """
    Plots a clean bar chart comparing Baseline vs Robust under combined stress (Level 5)
    """

    # Filter only combined level 5
    subset = results_df[
        (results_df["noise_type"] == "combined") &
        (results_df["level"] == 5)
    ]

    if subset.empty:
        print("No combined level 5 data found.")
        return

    plt.figure(figsize=(8, 5))

    # For each model
    for model_name in subset["model"].unique():

        model_data = subset[subset["model"] == model_name]

        tracks = model_data["track"].values
        accs = model_data["test_acc"].values

        x = np.arange(len(tracks))

        plt.bar(
            x + 0.2 * list(subset["model"].unique()).index(model_name),
            accs,
            width=0.2,
            label=model_name
        )

    # X-axis labels (Baseline vs Robust)
    plt.xticks(
        ticks=np.arange(len(tracks)),
        labels=tracks
    )

    plt.ylabel("Accuracy (%)")
    plt.xlabel("Training Track")
    plt.title("Performance under Combined Level 5 Stress")

    plt.legend()
    plt.grid(axis='y', alpha=0.3)

    plt.tight_layout()

    save_path = os.path.join(RESULTS_DIR, "combined_bar_comparison.png")
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"Saved combined comparison plot to: {save_path}")
```
